```python
from pathlib import Path
import re

folder = Path("/home/mjirik/mnt/meta/auto/plzen4-ntis/projects/korpusy-public/ZDO/ZDO2025/pilsen_pigs_all/kidneys_by_highschool_students")

pattern = re.compile(r"Tx(\d{3})D[_-]Art[_ -]?(Left|Right)[_ -]?kidney", re.IGNORECASE)
files = sorted(folder.glob("*.nii.gz"))

seen = {}  # pro počítání výskytů

for f in files:
    m = pattern.search(f.name)
    if not m:
        print(f"❌ Preskakuji {f.name} (neodpovídá vzoru)")
        continue
    animal_id, side = m.groups()
    key = (animal_id, side.lower())
    seen[key] = seen.get(key, 0) + 1
    idx = seen[key]

    new_name = f"Tx{animal_id}D_Art-{side.capitalize()}_kidney-manstuann_{idx}.nii.gz"
    new_path = f.with_name(new_name)
    print(f"{f.name}  →  {new_name}")
    f.rename(new_path)
```
